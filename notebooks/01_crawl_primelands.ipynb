# Part 1: Prime Lands Web Crawler

Production-grade async BFS crawler for primelands.lk

**Features:**
- Breadth-First Search with BFS queue
- Playwright for JavaScript-rendered content
- Exponential backoff with retries
- robots.txt compliance
- Content hash deduplication
- Rate limiting

```python
import sys
from pathlib import Path
import json

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from prime_lands.config import load_config
from prime_lands.crawler.crawler import PrimeLandsCrawler
from prime_lands.logger import setup_logger

# Setup logging
setup_logger(level="INFO")

print("✓ Imports successful")
```

## Step 1: Load Configuration

```python
# Load config from project root
cfg = load_config(Path.cwd().parent / "config.yaml")

print(f"Base URL: {cfg.crawler.base_url}")
print(f"Max pages: {cfg.crawler.max_pages}")
print(f"Rate limit: {cfg.crawler.rate_limit_seconds}s")
print(f"Respect robots.txt: {cfg.crawler.respect_robots_txt}")
```

## Step 2: Initialize Crawler

```python
data_dir = Path.cwd().parent / "data"
data_dir.mkdir(exist_ok=True)

crawler = PrimeLandsCrawler(cfg.crawler, data_dir=data_dir)
print(f"✓ Crawler initialized, output will be saved to: {data_dir}")
```

```python
# Run async crawler
properties = await crawler.crawl()

print(f"\n✓ Crawl complete!")
print(f"Total properties crawled: {len(properties)}")
print(f"Output file: {data_dir / 'primelands_corpus.jsonl'}")
```

## Step 3: Execute Crawl

⚠️ **Note**: This may take 5-15 minutes depending on page count and site responsiveness.

## Step 4: Analyze Crawled Data

```python
# Show sample properties
import pandas as pd

# Load corpus
corpus_file = data_dir / "primelands_corpus.jsonl"
properties_data = []

with open(corpus_file, "r", encoding="utf-8") as f:
    for line in f:
        properties_data.append(json.loads(line))

df = pd.DataFrame(properties_data)

print(f"Total properties: {len(df)}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nSample data:")
df.head(3)
```

```python
# Statistics
print("Property Statistics:")
print(f"- Avg description length: {df['description'].str.len().mean():.0f} chars")
print(f"- Properties with price: {df['price'].astype(bool).sum()}")
print(f"- Properties with bedrooms: {df['bedrooms'].notna().sum()}")
print(f"- Avg amenities per property: {df['amenities'].str.len().mean():.1f}")
```

## Step 5: Verify RAG Text Generation

```python
# Show RAG-formatted text for a sample property
from prime_lands.crawler.models import PropertyDocument

sample_prop = PropertyDocument(**properties_data[0])
rag_text = sample_prop.to_rag_text()

print("RAG-formatted text (first 500 chars):")
print(rag_text[:500])
print(f"\n...\nTotal length: {len(rag_text)} characters")
```
